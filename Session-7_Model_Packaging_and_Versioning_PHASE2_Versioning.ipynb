## 3.1: Semantic Versioning (SemVer) for ML

### What is Semantic Versioning?

**Format:** `MAJOR.MINOR.PATCH` (e.g., `2.1.3`)

**Originally designed for software**, but we'll adapt it for ML models!

---

### Traditional SemVer (Software)

- **MAJOR:** Breaking changes (incompatible API changes)
- **MINOR:** New features (backwards-compatible)
- **PATCH:** Bug fixes (backwards-compatible)

**Example:**
- `1.0.0` → Initial release
- `1.0.1` → Bug fix
- `1.1.0` → New feature added
- `2.0.0` → Breaking change

---

### SemVer Adapted for ML Models

We need to adapt this for ML, where "breaking changes" and "features" mean different things!

**ML-Specific SemVer Rules:**

| Version Component | When to Increment | ML Examples |
|-------------------|-------------------|-------------|
| **MAJOR** (1.0.0 → 2.0.0) | Breaking changes that affect consumers | • New model architecture (RF → XGBoost)<br>• Different features (input schema changed)<br>• Different prediction format<br>• Different preprocessing<br>• Incompatible with previous API |
| **MINOR** (1.0.0 → 1.1.0) | Improvements without breaking changes | • Hyperparameter tuning<br>• Retrained on more/better data<br>• Better preprocessing (but same input)<br>• Performance improvements<br>• Same API, better results |
| **PATCH** (1.0.0 → 1.0.1) | Bug fixes and tiny improvements | • Fix inference code bugs<br>• Fix preprocessing errors<br>• Minor corrections<br>• No retraining, just fixes |

---

### 🎯 Decision Framework for ML Versioning

**Ask these questions to determine version bump:**

1. **Did the input schema change?**
   - Different features?
   - Different feature order?
   - Different data types?
   - **YES** → MAJOR bump

2. **Did the model architecture change?**
   - Different algorithm?
   - Different framework?
   - **YES** → MAJOR bump

3. **Did the output format change?**
   - Binary → Multi-class?
   - Probabilities → Labels?
   - **YES** → MAJOR bump

4. **Did you retrain with more/better data?**
   - Same features, more data?
   - Better data quality?
   - **YES** → MINOR bump

5. **Did you tune hyperparameters?**
   - Better performance, same architecture?
   - **YES** → MINOR bump

6. **Did you fix a bug without retraining?**
   - Preprocessing fix?
   - Inference code fix?
   - **YES** → PATCH bump

## 3.2: Git Integration for Model Versioning

### Why Sync Models with Code?

**The Problem:**
- Model v2.1.0 is deployed
- But which CODE trained it?
- Which COMMIT was used?
- Can you reproduce it?

**The Solution: Connect model versions to Git commits**

---

### Git Tags for Model Versions

**Git tags** create a permanent marker in your repository history.

**Best Practice Pattern:**
```bash
# After training a new model version
git add .
git commit -m "Train model v1.2.0 with hyperparameter tuning"
git tag -a v1.2.0 -m "Model v1.2.0: RF with n_estimators=200, max_depth=15"
git push origin v1.2.0
```

**Now you have:**
- Model artifact: `model_v1.2.0.pkl`
- Git tag: `v1.2.0`
- MLflow run: Linked to git commit hash
- Complete reproducibility! 🎯

---

### Tracking Git Metadata in Models

**What to capture:**
```python
import subprocess

def get_git_metadata():
    return {
        'commit_hash': subprocess.check_output(
            ['git', 'rev-parse', 'HEAD']
        ).decode().strip(),
        'branch': subprocess.check_output(
            ['git', 'rev-parse', '--abbrev-ref', 'HEAD']
        ).decode().strip(),
        'remote_url': subprocess.check_output(
            ['git', 'config', '--get', 'remote.origin.url']
        ).decode().strip(),
        'is_dirty': len(subprocess.check_output(
            ['git', 'status', '--porcelain']
        ).decode().strip()) > 0
    }
```

**Include this in your model metadata!**

````python
# Hands-On: Git Integration Helper

import subprocess
import json
from datetime import datetime

print("🔗 Git Integration for Model Versioning\n")
print("="*70)

def get_git_metadata():
    """
    Capture Git metadata for model versioning.
    
    Returns:
        dict: Git metadata including commit hash, branch, etc.
    """
    try:
        commit_hash = subprocess.check_output(
            ['git', 'rev-parse', 'HEAD'],
            stderr=subprocess.DEVNULL
        ).decode().strip()
        
        branch = subprocess.check_output(
            ['git', 'rev-parse', '--abbrev-ref', 'HEAD'],
            stderr=subprocess.DEVNULL
        ).decode().strip()
        
        # Check for uncommitted changes
        status = subprocess.check_output(
            ['git', 'status', '--porcelain'],
            stderr=subprocess.DEVNULL
        ).decode().strip()
        
        is_dirty = len(status) > 0
        
        # Get remote URL
        try:
            remote_url = subprocess.check_output(
                ['git', 'config', '--get', 'remote.origin.url'],
                stderr=subprocess.DEVNULL
            ).decode().strip()
        except:
            remote_url = None
        
        return {
            'commit_hash': commit_hash,
            'commit_hash_short': commit_hash[:7],
            'branch': branch,
            'remote_url': remote_url,
            'is_dirty': is_dirty,
            'warning': 'Uncommitted changes detected!' if is_dirty else None
        }
    except Exception as e:
        print(f"⚠️  Git not available or not a git repository")
        return {
            'commit_hash': None,
            'error': str(e)
        }

# Try to get git metadata
print("Attempting to capture Git metadata...\n")
git_metadata = get_git_metadata()

if git_metadata.get('commit_hash'):
    print("✅ Git metadata captured:")
    print(f"   Commit: {git_metadata['commit_hash_short']} ({git_metadata['commit_hash']})")
    print(f"   Branch: {git_metadata['branch']}")
    if git_metadata['remote_url']:
        print(f"   Remote: {git_metadata['remote_url']}")
    
    if git_metadata['is_dirty']:
        print(f"\n   ⚠️  WARNING: You have uncommitted changes!")
        print(f"      Best practice: Commit before training models")
else:
    print("ℹ️  Running outside git repository (demo environment)")
    print("   In production, always use Git!")

print("\n" + "="*70)

# Example: Complete model metadata with Git info
print("\n📋 Example: Complete Model Metadata\n")

complete_metadata = {
    "model_version": "1.2.0",
    "model_name": "customer_churn_classifier",
    "algorithm": "RandomForestClassifier",
    "trained_at": datetime.now().isoformat(),
    "trained_by": "data_scientist@company.com",
    
    # Git information
    "git": git_metadata,
    
    # Training information
    "training": {
        "dataset_version": "2024-11-01",
        "dataset_hash": "a3f2e9c8b7d6",
        "n_samples": 10000,
        "features": ["age", "tenure", "monthly_charges", "total_charges"]
    },
    
    # Performance metrics
    "metrics": {
        "accuracy": 0.8789,
        "precision": 0.8654,
        "recall": 0.8923,
        "f1_score": 0.8787
    },
    
    # Model artifacts
    "artifacts": {
        "model_path": "models/model_v1.2.0.pkl",
        "onnx_path": "models/model_v1.2.0.onnx",
        "requirements": "requirements_v1.2.0.txt"
    }
}

print(json.dumps(complete_metadata, indent=2))

print("\n" + "="*70)
print("\n💡 PRO TIPS:")
print("   1. Always commit code before training")
print("   2. Tag commits with model versions")
print("   3. Include git hash in model metadata")
print("   4. Warn if training with uncommitted changes")
print("   5. Link MLflow runs to git commits")
````

# 📦 Section 3.3: Metadata Tracking (Continued)
## Complete Model Provenance

### What Metadata Should Every Model Have?

Professional model packages include **complete metadata** that answers:
- **What** is this model?
- **Who** created it?
- **When** was it trained?
- **Where** is the training code?
- **Why** was it trained? (business objective)
- **How** was it trained? (data, hyperparameters, etc.)
- **How well** does it perform?

---

### 📋 Complete Metadata Checklist

**1. Model Identity**
- Model name
- Version (SemVer)
- Model type/architecture
- Framework and versions

**2. Training Information**
- Training timestamp
- Dataset version/hash
- Number of training samples
- Feature list (names, types)
- Train/validation/test split

**3. Code Provenance**
- Git commit hash
- Git branch
- Git repository URL
- Uncommitted changes flag

**4. Performance Metrics**
- Training metrics
- Validation metrics
- Test metrics
- Business metrics (if applicable)

**5. Hyperparameters**
- All hyperparameters used
- Random seeds
- Optimization settings

**6. Environment**
- Python version
- Library versions (exact!)
- Hardware used (CPU/GPU)
- OS information

**7. Model Artifacts**
- Model file paths
- Input/output schemas
- Preprocessing pipeline
- Related artifacts (plots, reports)

**8. Governance**
- Trained by (person/email)
- Approved by
- Business owner
- Compliance notes

**9. Documentation**
- Model purpose
- Limitations
- Known issues
- Usage instructions

````python
# Hands-On: Creating Complete Model Metadata

import json
import hashlib
import platform
from datetime import datetime
import sys
import sklearn
import numpy as np
import pandas as pd

print("📋 Creating Complete Model Metadata\n")
print("="*70)

def create_model_metadata(
    model_version,
    model_name,
    algorithm,
    hyperparameters,
    dataset_info,
    metrics,
    trained_by,
    git_metadata=None
):
    """
    Create comprehensive model metadata.
    
    This is a PRODUCTION-GRADE metadata template!
    """
    metadata = {
        # === IDENTITY ===
        "model_identity": {
            "name": model_name,
            "version": model_version,
            "algorithm": algorithm,
            "framework": "scikit-learn",
            "framework_version": sklearn.__version__,
            "model_type": "classification",  # or "regression", "clustering", etc.
            "task": "binary_classification",  # specific task
        },
        
        # === TRAINING INFO ===
        "training": {
            "trained_at": datetime.now().isoformat(),
            "trained_by": trained_by,
            "training_duration_seconds": 45.2,  # Track this during training
            "dataset": dataset_info,
            "hyperparameters": hyperparameters,
            "random_seed": hyperparameters.get("random_state", None),
        },
        
        # === CODE PROVENANCE ===
        "provenance": {
            "git": git_metadata or {},
            "mlflow_run_id": "abc123def456",  # Link to MLflow
            "mlflow_experiment_id": "2",
        },
        
        # === PERFORMANCE ===
        "performance": {
            "metrics": metrics,
            "validation_date": datetime.now().isoformat(),
            "metric_confidence": "high",  # based on sample size
        },
        
        # === ENVIRONMENT ===
        "environment": {
            "python_version": f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}",
            "platform": platform.platform(),
            "processor": platform.processor(),
            "libraries": {
                "sklearn": sklearn.__version__,
                "numpy": np.__version__,
                "pandas": pd.__version__,
            },
            "hardware": {
                "cpu_count": 8,  # Get from system
                "memory_gb": 16,
                "gpu": None,  # or GPU info
            }
        },
        
        # === ARTIFACTS ===
        "artifacts": {
            "model_file": f"model_{model_version}.pkl",
            "onnx_file": f"model_{model_version}.onnx",
            "requirements_file": f"requirements_{model_version}.txt",
            "conda_env_file": f"conda_{model_version}.yaml",
            "signature_file": f"signature_{model_version}.json",
            "metadata_file": f"metadata_{model_version}.json",
        },
        
        # === SCHEMA ===
        "schema": {
            "input_features": dataset_info.get("features", []),
            "output_type": "binary_probability",
            "input_shape": [len(dataset_info.get("features", []))],
            "output_shape": [2],  # binary classification
        },
        
        # === GOVERNANCE ===
        "governance": {
            "business_owner": "product_team@company.com",
            "approved_by": None,  # Set when approved
            "approval_date": None,
            "compliance_notes": "PII-free model",
            "regulatory_requirements": [],
        },
        
        # === DOCUMENTATION ===
        "documentation": {
            "purpose": "Predict customer churn probability",
            "use_cases": [
                "Production real-time inference",
                "Batch scoring for retention campaigns"
            ],
            "limitations": [
                "Trained on US customers only",
                "May underperform on new customer segments",
                "Requires monthly retraining"
            ],
            "known_issues": [],
        },
        
        # === VERSIONING ===
        "versioning": {
            "previous_version": "1.1.0",
            "changes_from_previous": "Hyperparameter tuning improved accuracy by 2.2%",
            "breaking_changes": False,
            "deprecation_date": None,
        }
    }
    
    return metadata

# Example usage
print("Creating metadata for customer churn model v1.2.0...\n")

metadata = create_model_metadata(
    model_version="1.2.0",
    model_name="customer_churn_classifier",
    algorithm="RandomForestClassifier",
    hyperparameters={
        "n_estimators": 200,
        "max_depth": 15,
        "min_samples_split": 5,
        "min_samples_leaf": 2,
        "random_state": 42,
    },
    dataset_info={
        "version": "2024-11-01",
        "hash": "a3f2e9c8b7d6f1a4e5b9c2d8",
        "n_samples": 10000,
        "n_features": 4,
        "features": ["age", "tenure", "monthly_charges", "total_charges"],
        "target": "churn",
        "split": {"train": 0.8, "val": 0.1, "test": 0.1},
    },
    metrics={
        "train": {
            "accuracy": 0.8921,
            "precision": 0.8834,
            "recall": 0.9001,
            "f1": 0.8917,
            "auc_roc": 0.9234,
        },
        "validation": {
            "accuracy": 0.8789,
            "precision": 0.8654,
            "recall": 0.8923,
            "f1": 0.8787,
            "auc_roc": 0.9101,
        },
        "test": {
            "accuracy": 0.8812,
            "precision": 0.8698,
            "recall": 0.8945,
            "f1": 0.8820,
            "auc_roc": 0.9145,
        }
    },
    trained_by="amey.ds@company.com",
    git_metadata={
        "commit_hash": "7b4c1f2",
        "branch": "main",
        "remote_url": "github.com/company/ml-models",
        "is_dirty": False,
    }
)

# Display metadata (pretty printed)
print(json.dumps(metadata, indent=2))

# Save metadata to file
metadata_file = "metadata_v1.2.0.json"
with open(metadata_file, 'w') as f:
    json.dump(metadata, f, indent=2)

print(f"\n{'='*70}")
print(f"✅ Metadata saved to: {metadata_file}")
print(f"\n💡 This metadata file will travel with your model everywhere!")
print(f"   Anyone can load it and understand EVERYTHING about this model.")
````
