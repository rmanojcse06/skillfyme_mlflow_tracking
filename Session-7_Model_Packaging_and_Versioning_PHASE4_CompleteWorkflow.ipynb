### Step-by-Step Packaging Process

**We'll build a complete package through these steps:**

**Let's do it!** 💪

````python
# Setup: Import all necessary libraries

import os
import json
import joblib
import numpy as np
import pandas as pd
import random
import sys
import subprocess
import shutil
from datetime import datetime
from pathlib import Path

# ML libraries
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# MLflow
import mlflow
from mlflow.models import infer_signature

print("✅ All libraries imported successfully!")
print("\n🎯 Ready to build a production-grade model package!")
````

````python
# Step 1: Set up reproducibility (from Section 4)

print("🌱 Step 1: Setting Up Reproducibility\n")
print("="*70)

def set_all_seeds(seed=42):
    """Set all random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    print(f"✅ All seeds set to: {seed}")
    return seed

# Set seed
RANDOM_SEED = 42
set_all_seeds(RANDOM_SEED)

# Define version
MODEL_VERSION = "1.0.0"  # Our first production model!
MODEL_NAME = "customer_churn_classifier"

print(f"\n📦 Model: {MODEL_NAME}")
print(f"🏷️  Version: {MODEL_VERSION}")
print(f"🌱 Seed: {RANDOM_SEED}")
````

````python
# Step 2: Train model (simulating Session 6 workflow)

print("\n" + "="*70)
print("\n🤖 Step 2: Training Model\n")

# Load data (using iris as example, but imagine this is your churn data)
data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name='target')

print(f"📊 Dataset loaded:")
print(f"   Features: {X.shape[1]}")
print(f"   Samples: {X.shape[0]}")
print(f"   Target classes: {len(np.unique(y))}")

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED, stratify=y
)

print(f"\n📊 Data split:")
print(f"   Training: {X_train.shape[0]} samples")
print(f"   Testing: {X_test.shape[0]} samples")

# Train model
print("\n⏳ Training Random Forest...")

model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=RANDOM_SEED,
    n_jobs=-1
)

start_time = datetime.now()
model.fit(X_train, y_train)
training_duration = (datetime.now() - start_time).total_seconds()

print(f"✅ Model trained in {training_duration:.2f} seconds")

# Evaluate
print("\n📊 Evaluating model...")

y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

train_metrics = {
    'accuracy': accuracy_score(y_train, y_pred_train),
    'precision': precision_score(y_train, y_pred_train, average='weighted'),
    'recall': recall_score(y_train, y_pred_train, average='weighted'),
    'f1': f1_score(y_train, y_pred_train, average='weighted')
}

test_metrics = {
    'accuracy': accuracy_score(y_test, y_pred_test),
    'precision': precision_score(y_test, y_pred_test, average='weighted'),
    'recall': recall_score(y_test, y_pred_test, average='weighted'),
    'f1': f1_score(y_test, y_pred_test, average='weighted')
}

print("\n📈 Training Metrics:")
for metric, value in train_metrics.items():
    print(f"   {metric}: {value:.4f}")

print("\n📈 Test Metrics:")
for metric, value in test_metrics.items():
    print(f"   {metric}: {value:.4f}")

print("\n✅ Model training complete!")
````

````python
# Step 3: Create package directory structure

print("\n" + "="*70)
print("\n📁 Step 3: Creating Package Structure\n")

# Create base directory
package_name = f"{MODEL_NAME}_v{MODEL_VERSION}"
package_dir = Path(package_name)

# Create subdirectories
subdirs = [
    'models',
    'environment',
    'metadata',
    'code',
    'docs',
    'examples'
]

# Remove if exists (for clean demo)
if package_dir.exists():
    shutil.rmtree(package_dir)

# Create directories
package_dir.mkdir(exist_ok=True)
for subdir in subdirs:
    (package_dir / subdir).mkdir(exist_ok=True)
    print(f"✅ Created: {package_name}/{subdir}/")

print(f"\n📦 Package structure created at: {package_dir.absolute()}")
````

````python
# Step 4: Save models in multiple formats

print("\n" + "="*70)
print("\n💾 Step 4: Saving Models in Multiple Formats\n")

# 4a: Save as joblib (recommended for sklearn)
joblib_path = package_dir / 'models' / 'model.joblib'
joblib.dump(model, joblib_path)
joblib_size = joblib_path.stat().st_size
print(f"✅ Saved joblib: {joblib_path.name} ({joblib_size:,} bytes)")

# 4b: Save as pickle (for compatibility)
import pickle
pickle_path = package_dir / 'models' / 'model.pkl'
with open(pickle_path, 'wb') as f:
    pickle.dump(model, f)
pickle_size = pickle_path.stat().st_size
print(f"✅ Saved pickle: {pickle_path.name} ({pickle_size:,} bytes)")

# 4c: Save as ONNX (for production deployment)
print("\n⏳ Converting to ONNX...")
try:
    from skl2onnx import convert_sklearn
    from skl2onnx.common.data_types import FloatTensorType
    
    # Define input type
    initial_type = [('float_input', FloatTensorType([None, X_train.shape[1]]))]
    
    # Convert
    onnx_model = convert_sklearn(model, initial_types=initial_type, target_opset=12)
    
    # Save
    onnx_path = package_dir / 'models' / 'model.onnx'
    with open(onnx_path, 'wb') as f:
        f.write(onnx_model.SerializeToString())
    
    onnx_size = onnx_path.stat().st_size
    print(f"✅ Saved ONNX: {onnx_path.name} ({onnx_size:,} bytes)")
except ImportError:
    print("⚠️  skl2onnx not available (install with: pip install skl2onnx)")
    print("   Skipping ONNX conversion (not critical for demo)")

print("\n📊 Model Serialization Summary:")
print(f"   pickle:  {pickle_size:,} bytes")
print(f"   joblib:  {joblib_size:,} bytes (recommended)")
try:
    print(f"   ONNX:    {onnx_size:,} bytes (production deployment)")
except:
    pass
````

````python
# Step 5: Generate environment specifications

print("\n" + "="*70)
print("\n🔧 Step 5: Generating Environment Specifications\n")

# 5a: requirements.txt
requirements = [
    f"scikit-learn=={__import__('sklearn').__version__}",
    f"numpy=={np.__version__}",
    f"pandas=={pd.__version__}",
    f"joblib=={joblib.__version__}",
    "mlflow>=2.9.0",
]

req_path = package_dir / 'environment' / 'requirements.txt'
with open(req_path, 'w') as f:
    f.write('\n'.join(requirements))

print("✅ Generated requirements.txt:")
for req in requirements:
    print(f"   {req}")

# 5b: conda.yaml
conda_yaml = f"""
name: {MODEL_NAME}_v{MODEL_VERSION.replace('.', '_')}
channels:
  - defaults
  - conda-forge
dependencies:
  - python={sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}
  - pip
  - pip:
    - scikit-learn=={__import__('sklearn').__version__}
    - numpy=={np.__version__}
    - pandas=={pd.__version__}
    - joblib=={joblib.__version__}
    - mlflow>=2.9.0
""".strip()

conda_path = package_dir / 'environment' / 'conda.yaml'
with open(conda_path, 'w') as f:
    f.write(conda_yaml)

print("\n✅ Generated conda.yaml")

# 5c: Python version file
python_version_path = package_dir / 'environment' / 'python_version.txt'
with open(python_version_path, 'w') as f:
    f.write(f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}")

print(f"✅ Generated python_version.txt: {sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}")
````

````python
# Step 6: Create comprehensive metadata

print("\n" + "="*70)
print("\n📋 Step 6: Creating Complete Metadata\n")

# Get git metadata (if available)
def get_git_metadata():
    try:
        commit = subprocess.check_output(
            ['git', 'rev-parse', 'HEAD'], 
            stderr=subprocess.DEVNULL
        ).decode().strip()
        branch = subprocess.check_output(
            ['git', 'rev-parse', '--abbrev-ref', 'HEAD'],
            stderr=subprocess.DEVNULL
        ).decode().strip()
        return {
            'commit_hash': commit,
            'commit_hash_short': commit[:7],
            'branch': branch
        }
    except:
        return {'commit_hash': 'not_available', 'branch': 'not_available'}

git_info = get_git_metadata()

# Create complete metadata
metadata = {
    "model_identity": {
        "name": MODEL_NAME,
        "version": MODEL_VERSION,
        "algorithm": "RandomForestClassifier",
        "framework": "scikit-learn",
        "framework_version": __import__('sklearn').__version__,
    },
    "training": {
        "trained_at": datetime.now().isoformat(),
        "trained_by": "data_scientist@company.com",
        "training_duration_seconds": training_duration,
        "random_seed": RANDOM_SEED,
        "dataset": {
            "name": "iris",
            "n_samples": len(X),
            "n_features": X.shape[1],
            "features": list(X.columns),
            "target": "target",
            "n_classes": len(np.unique(y)),
        },
        "hyperparameters": {
            "n_estimators": 100,
            "max_depth": 10,
            "min_samples_split": 5,
            "min_samples_leaf": 2,
            "random_state": RANDOM_SEED,
        },
    },
    "performance": {
        "train": train_metrics,
        "test": test_metrics,
    },
    "provenance": {
        "git": git_info,
    },
    "environment": {
        "python_version": f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}",
        "scikit_learn": __import__('sklearn').__version__,
        "numpy": np.__version__,
        "pandas": pd.__version__,
    },
    "artifacts": {
        "model_joblib": "models/model.joblib",
        "model_pickle": "models/model.pkl",
        "model_onnx": "models/model.onnx",
        "requirements": "environment/requirements.txt",
        "conda_env": "environment/conda.yaml",
    },
}

# Save metadata
metadata_path = package_dir / 'metadata' / 'metadata.json'
with open(metadata_path, 'w') as f:
    json.dump(metadata, f, indent=2)

print("✅ Created metadata.json with complete provenance")
print("\n📄 Metadata includes:")
for key in metadata.keys():
    print(f"   • {key}")
````

````python
# Step 7: Create model signature

print("\n" + "="*70)
print("\n📐 Step 7: Creating Model Signature\n")

# Infer signature from training data
signature = infer_signature(X_train, y_pred_train)

# Convert to dict for JSON storage
signature_dict = {
    "inputs": [
        {"name": col, "type": "double"}
        for col in X_train.columns
    ],
    "outputs": [
        {"type": "long", "description": "Predicted class (0, 1, or 2)"}
    ],
    "example_input": X_test.iloc[0].to_dict(),
    "example_output": int(y_pred_test[0]),
}

signature_path = package_dir / 'metadata' / 'signature.json'
with open(signature_path, 'w') as f:
    json.dump(signature_dict, f, indent=2)

print("✅ Created signature.json")
print(f"\n📊 Input features ({len(signature_dict['inputs'])}):")
for inp in signature_dict['inputs']:
    print(f"   • {inp['name']} ({inp['type']})")
print(f"\n📊 Output: {signature_dict['outputs'][0]['description']}")
````

````python
# Step 10: Final package summary

print("\n" + "="*70)
print("\n🎉 Step 10: Package Complete!\n")
print("="*70)

# Show package structure
print(f"\n📦 Package: {package_name}\n")

def show_tree(directory, prefix="", max_depth=3, current_depth=0):
    """Display directory tree."""
    if current_depth >= max_depth:
        return
    
    items = sorted(directory.iterdir())
    for i, item in enumerate(items):
        is_last = i == len(items) - 1
        print(f"{prefix}{'└── ' if is_last else '├── '}{item.name}")
        
        if item.is_dir():
            extension = "    " if is_last else "│   "
            show_tree(item, prefix + extension, max_depth, current_depth + 1)

show_tree(package_dir)

# Calculate package size
total_size = sum(f.stat().st_size for f in package_dir.rglob('*') if f.is_file())
print(f"\n📊 Package Statistics:")
print(f"   Total size: {total_size:,} bytes ({total_size/1024:.2f} KB)")
print(f"   Files: {len(list(package_dir.rglob('*')))}")
print(f"   Version: {MODEL_VERSION}")
print(f"   Random seed: {RANDOM_SEED}")

print(f"\n✅ COMPLETE MODEL PACKAGE CREATED!")
print(f"\n📂 Location: {package_dir.absolute()}")
print(f"\n🚀 This package is now ready for:")
print(f"   ✅ Version control (Git)")
print(f"   ✅ MLflow registration")
print(f"   ✅ Docker containerization")
print(f"   ✅ Kubernetes deployment")
print(f"   ✅ Cloud platform upload")
print(f"   ✅ Production serving")

print("\n" + "="*70)
````

## 5.3: MLflow Model Format

### What is MLflow's Native Format?

MLflow has its own packaging standard that includes everything we just created, PLUS:
- Multiple "flavors" (sklearn, pytorch, tensorflow, etc.)
- Automatic signature inference
- Built-in serving capabilities
- Version management integration

**MLflow Package Structure:**
```
model/
├── MLmodel              # MLflow metadata file
├── conda.yaml           # Environment spec
├── requirements.txt     # Python dependencies
├── python_model.pkl     # Serialized model
└── model.pkl            # Or framework-specific format
```

**The MLmodel file:**
```yaml
flavors:
  python_function:
    env: conda.yaml
    loader_module: mlflow.sklearn
    model_path: model.pkl
    python_version: 3.11.0
  sklearn:
    pickled_model: model.pkl
    serialization_format: cloudpickle
    sklearn_version: 1.3.0
```

---

### Manual Package vs MLflow Package

**Our Manual Package:**
- ✅ Complete control
- ✅ Custom structure
- ✅ Multiple formats (ONNX, etc.)
- ✅ Detailed documentation
- ❌ More work to create
- ❌ Custom loading code needed

**MLflow Package:**
- ✅ Automatic creation
- ✅ Standard format
- ✅ One-line serving: `mlflow models serve`
- ✅ Registry integration
- ❌ Less flexibility
- ❌ Framework limitations

---

### 🎯 Best Practice: Use Both!

**Strategy:**
1. Create our custom package (full control, ONNX, docs)
2. ALSO log to MLflow (easy serving, registry)
3. Reference MLflow run in custom package metadata

**This gives you:**
- Custom packaging flexibility
