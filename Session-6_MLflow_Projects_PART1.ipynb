## 📖 Section 1: Introduction to MLflow Projects (15 min)

### The Problem: Reproducibility Crisis in ML

Imagine this scenario:

### 🎬 Real-World Scenario

**You (Data Scientist):**  
"Hey team, I trained an amazing model! 87.5% accuracy!"  
*Shares: `train_model.py`*

**Teammate tries to run it:**
```bash
python train_model.py
```

**Result:**
```
❌ ModuleNotFoundError: No module named 'xgboost'
❌ FileNotFoundError: data/train.csv
❌ Different results: 79.2% accuracy (WTF?!)
```

**Why it failed:**
- ❌ Different package versions
- ❌ Missing dependencies
- ❌ Unknown data location
- ❌ Different random seeds
- ❌ No documentation

**This happens EVERY DAY in ML teams!**

### 💡 The Solution: MLflow Projects

MLflow Projects solve the "works on my machine" problem by providing:

1. **Standard Structure** - Everyone knows where files are
2. **Dependency Management** - Exact package versions
3. **Entry Points** - Clear interfaces for running code
4. **Parameters** - Configurable without code changes
5. **Reproducibility** - Same code, same results, anywhere

---

### 🎯 What is an MLflow Project?

An MLflow Project is a **standardized format** for packaging ML code.

**Minimum Requirements:**
1. Code files (Python scripts)
2. `MLproject` file (project definition)
3. `conda.yaml` or `requirements.txt` (dependencies)

**Optional but Recommended:**
- README.md (documentation)
- Data files or instructions
- Configuration files

**Think of it as:** A recipe for your ML experiment that anyone can follow!

### 🏆 Benefits of MLflow Projects

| Problem | Solution with Projects |
|---------|------------------------|
| "Works on my machine only" | Reproducible everywhere |
| "What packages did you use?" | Dependencies declared |
| "How do I run this?" | Clear entry points |
| "Can I change hyperparameters?" | Parameterized runs |
| "No documentation" | Self-documenting structure |
| "Can't share code" | Git integration |

---

### 🎓 When to Use MLflow Projects

**✅ Use Projects When:**
- Sharing code with teammates
- Running experiments on different machines
- Need reproducible results
- Automating ML workflows
- Building production pipelines

**❌ Don't Need Projects For:**
- Quick Jupyter notebook exploration
- One-time throwaway scripts
- Personal experiments (though still helpful!)

**Pro Tip:** If you'll run it more than once or share it with anyone, make it a Project!

## 📖 Section 2: Anatomy of MLflow Projects (15 min)

### 📁 Project Structure

A typical MLflow Project looks like this:

```
my_ml_project/
├── MLproject          # Project definition (REQUIRED)
├── conda.yaml         # Environment spec (REQUIRED)
├── train.py           # Your ML code
├── evaluate.py        # Evaluation code
├── README.md          # Documentation
└── data/              # Data files (optional)
```

Let's understand each component!

### 📄 Component 1: The MLproject File

The `MLproject` file is the **heart** of your project. It's a YAML file that defines:
- Project name
- Environment (conda or docker)
- Entry points (how to run your code)
- Parameters (what can be configured)

**Example MLproject File:**

```yaml
name: customer_churn_classifier

conda_env: conda.yaml

entry_points:
  main:
    parameters:
      n_estimators: {type: int, default: 100}
      max_depth: {type: int, default: 10}
      data_path: {type: str, default: "data/train.csv"}
    command: "python train.py {n_estimators} {max_depth} {data_path}"
```

**Let's break this down!**

### 🔍 MLproject File Components

#### 1. **name** - Project identifier
```yaml
name: customer_churn_classifier
```
- Short, descriptive name
- Use underscores, not spaces
- Should describe what the project does

---

#### 2. **conda_env** - Environment specification
```yaml
conda_env: conda.yaml
```
- Points to environment file
- Can also use `docker_env` for containers
- Ensures same package versions everywhere

---

#### 3. **entry_points** - How to run your code

Entry points are like **functions** you can call:

```yaml
entry_points:
  main:           # Default entry point
    parameters:   # What can be configured
      n_estimators: {type: int, default: 100}
    command: "python train.py {n_estimators}"
  
  evaluate:       # Another entry point
    parameters:
      model_path: {type: str}
    command: "python evaluate.py {model_path}"
```

**Multiple entry points = Different ways to use your project!**

### 📋 Parameter Types

Parameters in MLproject files can have different types:

| Type | Example | Usage |
|------|---------|-------|
| `int` | `n_estimators: {type: int, default: 100}` | Integers |
| `float` | `learning_rate: {type: float, default: 0.01}` | Decimals |
| `str` | `data_path: {type: str, default: "data.csv"}` | Strings |
| `path` | `model_path: {type: path}` | File paths |

**Best Practices:**
- ✅ Always provide defaults
- ✅ Use descriptive parameter names
- ✅ Group related parameters
- ✅ Document what each parameter does (in README)

---

### 🎯 Command String

The `command` tells MLflow how to run your code:

```yaml
command: "python train.py {n_estimators} {max_depth}"
```

**How it works:**
```bash
# You run:
mlflow run . -P n_estimators=200 -P max_depth=15

# MLflow executes:
python train.py 200 15
```

**Pro Tip:** Use argparse in your Python scripts to handle these arguments!

### 📄 Component 2: The conda.yaml File

The `conda.yaml` file specifies your **environment** - all dependencies and versions.

**Example conda.yaml:**

```yaml
name: churn_classifier_env
channels:
  - conda-forge
  - defaults
dependencies:
  - python=3.10
  - pip
  - pip:
      - mlflow==2.9.2
      - scikit-learn==1.3.2
      - pandas==2.1.4
      - numpy==1.24.3
```

**Why this matters:**
- 🔒 **Lock versions** - Same results everywhere
- 🔄 **Reproducible** - Anyone can recreate environment
- 🚀 **Isolated** - No conflicts with other projects

**Pro Tip:** Use the EXACT versions from your Session 5 environment!

### 📁 Step 1: Create Project Directory Structure

First, let's create our project folder:

````python
import os
from pathlib import Path

# Create project directory
project_dir = Path("my_first_mlflow_project")
project_dir.mkdir(exist_ok=True)

print("✅ Created project directory:")
print(f"   {project_dir.absolute()}")

# We'll create files in this directory
os.chdir(project_dir)
print(f"\n📁 Current directory: {Path.cwd()}")
````

### 📝 Step 2: Create the MLproject File

Let's create our MLproject file with a simple entry point:

````python
mlproject_content = """name: my_first_mlflow_project

conda_env: conda.yaml

entry_points:
  main:
    parameters:
      n_estimators: {type: int, default: 100}
      max_depth: {type: int, default: 10}
      random_state: {type: int, default: 42}
    command: "python train.py --n-estimators {n_estimators} --max-depth {max_depth} --random-state {random_state}"
"""

# Write MLproject file
with open("MLproject", "w") as f:
    f.write(mlproject_content)

print("✅ Created MLproject file")
print("\n📄 Contents:")
print(mlproject_content)
````

### 🔍 Let's Understand What We Just Created

**Line by line breakdown:**

```yaml
name: my_first_mlflow_project    # Project identifier
```
↳ This is what shows up when you list projects

```yaml
conda_env: conda.yaml             # Environment file
```
↳ MLflow will create conda env from this file

```yaml
entry_points:
  main:                            # Default entry point
```
↳ When you run `mlflow run .`, this executes

```yaml
    parameters:
      n_estimators: {type: int, default: 100}
```
↳ Configurable via CLI: `-P n_estimators=200`

```yaml
    command: "python train.py --n-estimators {n_estimators} ..."
```
↳ This is what actually runs!

### 📝 Step 3: Create the conda.yaml File

Now let's specify our dependencies:

````python
conda_yaml_content = """name: mlflow_project_env
channels:
  - conda-forge
  - defaults
dependencies:
  - python=3.10
  - pip
  - pip:
      - mlflow==2.9.2
      - scikit-learn==1.3.2
      - pandas==2.1.4
      - numpy==1.24.3
"""

# Write conda.yaml file
with open("conda.yaml", "w") as f:
    f.write(conda_yaml_content)

print("✅ Created conda.yaml file")
print("\n📄 Contents:")
print(conda_yaml_content)

print("\n💡 Pro Tip: These are the SAME versions from Session 5!")
print("   This ensures reproducibility.")
````

### 🐍 Step 4: Create the Training Script

Now the actual ML code - this is adapted from your Session 5 code:

````python
train_py_content = '''"""Training script for MLflow Project
Adapted from Session 5 Random Forest code
"""

import argparse
import mlflow
import mlflow.sklearn
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

def train_model(n_estimators, max_depth, random_state):
    """Train Random Forest classifier"""
    
    print("🎯 Training Random Forest Classifier")
    print(f"   n_estimators: {n_estimators}")
    print(f"   max_depth: {max_depth}")
    print(f"   random_state: {random_state}")
    
    # Generate synthetic data (replace with your data)
    print("📊 Loading data...")
    X, y = make_classification(n_samples=1000, n_features=20, random_state=random_state)
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=random_state)
    print(f"   Train: {len(X_train)} samples")
    print(f"   Test: {len(X_test)} samples")
    
    # Train model
    print("🌲 Training model...")
    model = RandomForestClassifier(
        n_estimators=n_estimators,
        max_depth=max_depth,
        random_state=random_state,
        n_jobs=-1
    )
    model.fit(X_train, y_train)
    
    # Evaluate
    print("📈 Evaluating...")
    y_pred = model.predict(X_test)
    
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    
    print(f"   Accuracy: {accuracy:.4f}")
    print(f"   Precision: {precision:.4f}")
    print(f"   Recall: {recall:.4f}")
    print(f"   F1-Score: {f1:.4f}")
    
    # Log to MLflow
    print("📝 Logging to MLflow...")
    mlflow.log_param("n_estimators", n_estimators)
    mlflow.log_param("max_depth", max_depth)
    mlflow.log_param("random_state", random_state)
    
    mlflow.log_metric("accuracy", accuracy)
    mlflow.log_metric("precision", precision)
    mlflow.log_metric("recall", recall)
    mlflow.log_metric("f1_score", f1)
    
    mlflow.sklearn.log_model(model, "model")
    
    print("✅ Model logged to MLflow!")
    
    return model, {"accuracy": accuracy, "f1_score": f1}

if __name__ == "__main__":
    # Parse arguments
    parser = argparse.ArgumentParser()
    parser.add_argument("--n-estimators", type=int, default=100)
    parser.add_argument("--max-depth", type=int, default=10)
    parser.add_argument("--random-state", type=int, default=42)
    args = parser.parse_args()
    
    # Start MLflow run
    with mlflow.start_run():
        train_model(args.n_estimators, args.max_depth, args.random_state)
    
    print("🎉 Training complete!")
'''

# Write train.py file
with open("train.py", "w", encoding='utf-8') as f:
    f.write(train_py_content)

print("✅ Created train.py")
print("\n💡 This is your Session 5 code, but parameterized!")
````

## 🚀 Section 4: Running MLflow Projects (20 min)

### 🎯 Time to Run Your Project!

Now comes the exciting part - running your MLflow Project!

**Three ways to run:**
1. **Default parameters** - Quick start
2. **Custom parameters** - Tune hyperparameters
3. **Different entry points** - Run specific tasks

Let's try each!

### 🏃 Method 1: Run with Default Parameters

The simplest way to run your project:

````python
# Let's run the project!
# Note: In real terminal, you'd run: mlflow run .
# Here we'll use Python API

import mlflow

print("🚀 Running MLflow Project with default parameters...")
print("\nThis will:")
print("  1. Create conda environment")
print("  2. Install dependencies")
print("  3. Run training script")
print("  4. Log to MLflow")

# Run the project
submitted_run = mlflow.projects.run(
    uri=".",  # Current directory
    entry_point="main",  # Use main entry point
    # No parameters = use defaults!
)

print(f"\n✅ Project run submitted!")
print(f"   Run ID: {submitted_run.run_id}")
````

### 🎛️ Method 2: Run with Custom Parameters

Now let's customize the hyperparameters:

````python
print("🎯 Running with CUSTOM parameters...")
print("\nChanging from defaults:")
print("  n_estimators: 100 → 200")
print("  max_depth: 10 → 15")

# Run with custom parameters
submitted_run = mlflow.projects.run(
    uri=".",
    entry_point="main",
    parameters={
        "n_estimators": 200,
        "max_depth": 15,
        "random_state": 42
    }
)

print(f"\n✅ Project run submitted with custom params!")
print(f"   Run ID: {submitted_run.run_id}")

print("\n💡 In terminal, this would be:")
print("   mlflow run . -P n_estimators=200 -P max_depth=15")
````

### 🔄 Running Multiple Configurations

The real power: easily run many experiments!

````python
print("🔬 Running hyperparameter sweep...")
print("\nTrying different n_estimators values:")

configurations = [
    {"n_estimators": 50, "max_depth": 10},
    {"n_estimators": 100, "max_depth": 10},
    {"n_estimators": 200, "max_depth": 10},
]

run_ids = []

for i, config in enumerate(configurations, 1):
    print(f"\n{i}. Running with n_estimators={config['n_estimators']}...")
    
    submitted_run = mlflow.projects.run(
        uri=".",
        entry_point="main",
        parameters=config
    )
    
    run_ids.append(submitted_run.run_id)
    print(f"   ✅ Run ID: {submitted_run.run_id[:8]}...")

print("\n🎉 All configurations complete!")
print(f"\nGenerated {len(run_ids)} runs:")
for i, run_id in enumerate(run_ids, 1):
    print(f"   {i}. {run_id[:16]}...")

print("\n💡 Now check MLflow UI to compare these runs!")
````

## 🎓 Section 5: Advanced Project Features (10 min)

### 🎯 Multiple Entry Points

Real projects have multiple ways to use them!

**Example: A project with train AND evaluate:**

### 📝 Enhanced MLproject with Multiple Entry Points

```yaml
name: advanced_ml_project

conda_env: conda.yaml

entry_points:
  # Entry point 1: Training
  main:
    parameters:
      n_estimators: {type: int, default: 100}
      max_depth: {type: int, default: 10}
      data_path: {type: str, default: "data/train.csv"}
    command: "python train.py {n_estimators} {max_depth} {data_path}"
  
  # Entry point 2: Evaluation
  evaluate:
    parameters:
      model_uri: {type: str}
      test_data: {type: str, default: "data/test.csv"}
    command: "python evaluate.py {model_uri} {test_data}"
  
  # Entry point 3: Hyperparameter tuning
  tune:
    parameters:
      max_trials: {type: int, default: 10}
    command: "python tune.py {max_trials}"
```

**Usage:**
```bash
# Train model
mlflow run . -e main

# Evaluate specific model
mlflow run . -e evaluate -P model_uri="runs:/abc123/model"

# Run hyperparameter tuning
mlflow run . -e tune -P max_trials=20
```

**Pro Tip:** MLflow automatically tags project runs so you can find them easily!

````python
# Let's query our project runs programmatically
from mlflow.tracking import MlflowClient

client = MlflowClient()

print("🔍 Searching for project runs...")

# Search for runs with project tag
runs = client.search_runs(
    experiment_ids=["0"],  # Default experiment
    filter_string="tags.`mlflow.project.entryPoint` = 'main'",
    max_results=10
)

print(f"\n📊 Found {len(runs)} project runs:\n")

for i, run in enumerate(runs, 1):
    params = run.data.params
    metrics = run.data.metrics
    
    print(f"{i}. Run {run.info.run_id[:8]}...")
    print(f"   n_estimators: {params.get('n_estimators', 'N/A')}")
    print(f"   max_depth: {params.get('max_depth', 'N/A')}")
    print(f"   accuracy: {metrics.get('accuracy', 0):.4f}")
    print()

print("💡 These are all from your MLflow Projects!")
````

## 🎯 Section 8: Best Practices & Tips (10 min)

### ✅ MLflow Projects Best Practices

#### 1. **Keep Projects Focused**
```
❌ BAD: one_project_for_everything/
✅ GOOD: separate projects for each task
  - training_project/
  - evaluation_project/
  - deployment_prep_project/
```

#### 2. **Use Descriptive Names**
```yaml
❌ BAD: name: project1
✅ GOOD: name: customer_churn_rf_classifier
```

#### 3. **Always Provide Defaults**
```yaml
❌ BAD: 
  n_estimators: {type: int}

✅ GOOD:
  n_estimators: {type: int, default: 100}
```

#### 4. **Document Your Parameters**
Add a README.md explaining:
- What each parameter does
- Recommended value ranges
- Example usages

#### 5. **Pin Exact Versions**
```yaml
❌ BAD: 
  - scikit-learn

✅ GOOD:
  - scikit-learn==1.3.2
```

### 🚫 Common Mistakes to Avoid

#### Mistake 1: Forgetting conda.yaml
```
❌ Error: Environment specification not found
✅ Solution: Always create conda.yaml
```

#### Mistake 2: Hardcoding paths
```python
❌ BAD: data_path = "/Users/you/data/train.csv"
✅ GOOD: data_path = args.data_path  # From parameters
```

#### Mistake 3: Not using argparse
```python
❌ BAD: Hardcoded hyperparameters
✅ GOOD: Parse from command line
```

#### Mistake 4: Incomplete dependencies
```yaml
❌ BAD: Only listing main packages
✅ GOOD: List ALL dependencies (test on clean env)
```

### 💡 Pro Tips

#### Tip 1: Test Locally First
```bash
# Always test before pushing to Git
mlflow run . -P n_estimators=50  # Quick test
```

#### Tip 2: Use Environment Variables
```yaml
parameters:
  tracking_uri: {type: str, default: "${MLFLOW_TRACKING_URI}"}
```

#### Tip 3: Version Your Projects
```bash
git tag v1.0
git push --tags

# Others can use specific version
mlflow run https://github.com/you/project -v v1.0
```

#### Tip 4: Add Project Templates
Create template projects for common tasks:
- classification_template/
- regression_template/
- time_series_template/

Then copy and modify for new projects!
