### 🔑 Key Principles of Model Packaging

**1. Self-Contained**
Everything needed to run the model is included. No external dependencies on your local machine.

**2. Versioned**
Every package has a clear version number (v1.0.0, v1.1.0, v2.0.0) following semantic versioning.

**3. Reproducible**
Anyone can load the package and get the EXACT same predictions, every time.

**4. Documented**
Complete metadata explains what the model does, how it was trained, and how to use it.

**5. Validated**
Input/output schemas prevent invalid data from causing crashes.

**6. Portable**
Can be deployed to different platforms (cloud, edge, mobile) using framework-agnostic formats.

**7. Traceable**
Full lineage from data → code → training → model → deployment.

### Understanding Serialization

**Serialization** = Converting a model object (in memory) into a format that can be saved, transferred, and loaded later.

**Why Multiple Formats?**
Different deployment scenarios need different formats:
- Python-only deployment? → pickle/joblib
- Cross-language deployment? → ONNX
- Mobile/edge devices? → ONNX/TorchScript
- High-performance inference? → ONNX/TensorRT
- C++ production servers? → ONNX/TorchScript

**One size does NOT fit all!** Let's explore each format...

## 2.1: Python-Native Serialization

### pickle: Python's Built-in Serialization

**What is pickle?**
- Built into Python standard library
- Can serialize almost any Python object
- Simple to use

**When to use pickle:**
✅ Quick prototyping and experimentation  
✅ Python-only deployment  
✅ Simple models without complex dependencies  
✅ Short-term storage (days/weeks)  

**When NOT to use pickle:**
❌ Cross-language deployment (Java, C++, JavaScript)  
❌ Long-term storage (Python version changes)  
❌ Security-sensitive applications  
❌ Production systems (version compatibility issues)  
❌ Large models (poor compression)  

**Major Limitations:**
- ⚠️ **Version Compatibility:** Pickle from Python 3.9 may not work in Python 3.11
- ⚠️ **Security Risk:** Pickle can execute arbitrary code when loading (NEVER load untrusted pickle files!)
- ⚠️ **Python-Only:** Can't load pickle files in Java, C++, or other languages
- ⚠️ **Library Version Sensitivity:** sklearn 1.3 pickle won't load in sklearn 1.5

**Pro Tip:** Use pickle for experiments, NOT for production!

````python
# Hands-On: Using pickle

import pickle
import os
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

print("🔧 Demonstration: pickle Serialization\n")
print("="*60)

# Train a simple model
print("\n1️⃣ Training a simple Random Forest model...")
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)

model = RandomForestClassifier(n_estimators=10, random_state=42)
model.fit(X_train, y_train)
accuracy = model.score(X_test, y_test)
print(f"   ✅ Model trained! Accuracy: {accuracy:.4f}")

# Save with pickle
print("\n2️⃣ Saving model with pickle...")
pickle_path = 'demo_model.pkl'
with open(pickle_path, 'wb') as f:
    pickle.dump(model, f)

file_size = os.path.getsize(pickle_path)
print(f"   ✅ Model saved to {pickle_path}")
print(f"   📦 File size: {file_size:,} bytes ({file_size/1024:.2f} KB)")

# Load with pickle
print("\n3️⃣ Loading model with pickle...")
with open(pickle_path, 'rb') as f:
    loaded_model = pickle.load(f)

loaded_accuracy = loaded_model.score(X_test, y_test)
print(f"   ✅ Model loaded successfully!")
print(f"   🎯 Original accuracy: {accuracy:.4f}")
print(f"   🎯 Loaded accuracy: {loaded_accuracy:.4f}")
print(f"   ✅ Predictions match: {accuracy == loaded_accuracy}")

# Make predictions
print("\n4️⃣ Making predictions with loaded model...")
sample_prediction = loaded_model.predict([X_test[0]])
print(f"   📊 Sample prediction: {sample_prediction[0]} (True label: {y_test[0]})")

print("\n" + "="*60)
print("\n⚠️  IMPORTANT NOTES:")
print("   1. pickle is Python-version sensitive")
print("   2. sklearn version must match (training vs inference)")
print("   3. Never load pickle files from untrusted sources (security!)")
print("   4. Use pickle for experiments, not production")

# Cleanup
#os.remove(pickle_path)
#print("\n🧹 Demo file cleaned up")
````

### joblib: Optimized for Numerical Data

**What is joblib?**
- Created by scikit-learn developers
- Optimized for numpy arrays (common in ML)
- Better compression than pickle
- Faster for large numerical models

**When to use joblib:**
✅ scikit-learn models (officially recommended!)  
✅ Models with large numpy arrays  
✅ Models with many features  
✅ When file size matters  
✅ Python-only deployment  

**When NOT to use joblib:**
❌ Cross-language deployment  
❌ Deep learning models (use framework-specific formats)  
❌ Production systems requiring maximum compatibility  

**Advantages over pickle:**
- ⚡ **Better Compression:** 2-10x smaller files for numerical models
- ⚡ **Faster Load Times:** Optimized for large arrays
- ⚡ **Memory Efficient:** Uses memory mapping for large objects
- ✅ **Recommended by sklearn:** Official serialization format

**Still has pickle's limitations:**
- Python-version sensitivity
- Library version dependencies
- Security concerns (same as pickle)
- Python-only

**Pro Tip:** If using sklearn, prefer joblib over pickle!

````python
# Hands-On: joblib vs pickle comparison

import joblib
import pickle
import os
import time
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

print("🔧 Comparison: joblib vs pickle\n")
print("="*60)

# Train a larger model for comparison
print("\n1️⃣ Training a larger Random Forest model...")
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)

# Use more trees for bigger model
model = RandomForestClassifier(n_estimators=5000, max_depth=10, random_state=42)
model.fit(X_train, y_train)
print(f"   ✅ Model trained with 100 trees")

# Save with pickle
print("\n2️⃣ Saving with pickle...")
pickle_path = 'model_pickle_02.pkl'
start_time = time.time()
with open(pickle_path, 'wb') as f:
    pickle.dump(model, f)
pickle_save_time = time.time() - start_time
pickle_size = os.path.getsize(pickle_path)
print(f"   💾 Save time: {pickle_save_time:.4f} seconds")
print(f"   📦 File size: {pickle_size:,} bytes ({pickle_size/1024:.2f} KB)")

# Save with joblib
print("\n3️⃣ Saving with joblib...")
joblib_path = 'model_joblib_02.joblib'
start_time = time.time()
joblib.dump(model, joblib_path)
joblib_save_time = time.time() - start_time
joblib_size = os.path.getsize(joblib_path)
print(f"   💾 Save time: {joblib_save_time:.4f} seconds")
print(f"   📦 File size: {joblib_size:,} bytes ({joblib_size/1024:.2f} KB)")

# Load comparison
print("\n4️⃣ Loading comparison...")

# Load pickle
start_time = time.time()
with open(pickle_path, 'rb') as f:
    model_from_pickle = pickle.load(f)
pickle_load_time = time.time() - start_time
print(f"   ⏱️  pickle load time: {pickle_load_time:.4f} seconds")

# Load joblib
start_time = time.time()
model_from_joblib = joblib.load(joblib_path)
joblib_load_time = time.time() - start_time
print(f"   ⏱️  joblib load time: {joblib_load_time:.4f} seconds")

# Results comparison
print("\n" + "="*60)
print("📊 COMPARISON RESULTS:\n")
print(f"File Size:")
print(f"   pickle: {pickle_size/1024:.2f} KB")
print(f"   joblib: {joblib_size/1024:.2f} KB")
print(f"   💡 Compression: {((pickle_size-joblib_size)/pickle_size*100):.1f}% smaller with joblib")

print(f"\nSave Time:")
print(f"   pickle: {pickle_save_time:.4f}s")
print(f"   joblib: {joblib_save_time:.4f}s")

print(f"\nLoad Time:")
print(f"   pickle: {pickle_load_time:.4f}s")
print(f"   joblib: {joblib_load_time:.4f}s")

# Verify both work correctly
acc_pickle = model_from_pickle.score(X_test, y_test)
acc_joblib = model_from_joblib.score(X_test, y_test)
print(f"\nAccuracy Verification:")
print(f"   pickle model: {acc_pickle:.4f}")
print(f"   joblib model: {acc_joblib:.4f}")
print(f"   ✅ Both match: {acc_pickle == acc_joblib}")

print("\n" + "="*60)
print("\n🎯 KEY TAKEAWAY:")
print("   For sklearn models, joblib is the RECOMMENDED choice:")
print("   ✅ Smaller file size")
print("   ✅ Faster save/load (especially for large models)")
print("   ✅ Optimized for numerical data")
print("   ✅ Official sklearn recommendation")

# Cleanup
os.remove(pickle_path)
os.remove(joblib_path)
print("\n🧹 Demo files cleaned up")
````

## 2.2: Framework-Agnostic Formats

### The Cross-Platform Challenge

**Real-World Scenario:**

You trained a model in Python with scikit-learn. Now you need to:
- Deploy to a **C++ production server** (high performance)
- Run inference on **mobile devices** (Android/iOS)
- Integrate with a **Java microservice**
- Deploy to **edge devices** (IoT sensors)
- Use in a **JavaScript web application**

**Problem:** pickle and joblib only work in Python!

**Solution:** Framework-agnostic formats like ONNX and TorchScript

These formats allow you to:
1. Train in Python
2. Export to universal format
3. Deploy ANYWHERE (Python, C++, Java, JavaScript, mobile, edge)

**This is the future of ML deployment!** 🚀

### ONNX (Open Neural Network Exchange)

**What is ONNX?**
- Open standard for ML model representation
- Supported by: Microsoft, Facebook, AWS, NVIDIA, Intel, and more
- Works across frameworks: scikit-learn, PyTorch, TensorFlow, XGBoost, LightGBM
- Platform-agnostic: Python, C++, Java, JavaScript, C#, Go, mobile, edge

**When to use ONNX:**
✅ **Cross-language deployment** (Python → C++, Java, etc.)  
✅ **Production inference servers** (high performance)  
✅ **Mobile and edge deployment** (iOS, Android, IoT)  
✅ **Framework switching** (train in PyTorch, deploy in TensorFlow)  
✅ **Optimized inference** (ONNX Runtime is FAST!)  
✅ **Multi-platform support** (one model, many platforms)  

**When NOT to use ONNX:**
❌ Custom layers not yet supported  
❌ Rapid prototyping (conversion adds step)  
❌ Pure Python-only deployment  

**Major Advantages:**
- 🚀 **Framework Independence:** Train anywhere, deploy anywhere
- ⚡ **Optimized Inference:** ONNX Runtime is 2-10x faster than native frameworks
- 🌍 **Universal Support:** Works on cloud, mobile, edge, web
- 🔒 **Production Ready:** Battle-tested by major companies
- 📦 **Hardware Optimization:** Leverages GPUs, CPUs, specialized accelerators

**Supported Frameworks:**
- scikit-learn → ONNX (sklearn-onnx)
- PyTorch → ONNX (built-in)
- TensorFlow → ONNX (tf2onnx)
- XGBoost → ONNX (built-in)
- LightGBM → ONNX (built-in)
- Keras → ONNX (keras2onnx)

**Pro Tip:** ONNX is becoming the industry standard for production ML!

````python
# Hands-On: Converting sklearn model to ONNX

import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import joblib
import os
import time

# ONNX-specific imports
from skl2onnx import convert_sklearn
from skl2onnx.common.data_types import FloatTensorType
import onnxruntime as rt

print("🔧 Demonstration: Converting sklearn to ONNX\n")
print("="*70)

# Train a model
print("\n1️⃣ Training sklearn Random Forest...")
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42
)

model = RandomForestClassifier(n_estimators=10, random_state=42)
model.fit(X_train, y_train)
sklearn_acc = model.score(X_test, y_test)
print(f"   ✅ sklearn model trained! Accuracy: {sklearn_acc:.4f}")

# Make sklearn predictions
sklearn_pred = model.predict(X_test[:5])
sklearn_proba = model.predict_proba(X_test[:5])
print(f"   📊 Sample predictions: {sklearn_pred}")

# Convert to ONNX
print("\n2️⃣ Converting to ONNX format...")

# Define input type (4 features, float32)
initial_type = [('float_input', FloatTensorType([None, 4]))]

# Convert!
onnx_model = convert_sklearn(
    model, 
    initial_types=initial_type,
    target_opset=12  # ONNX opset version
)

# Save ONNX model
onnx_path = 'model.onnx'
with open(onnx_path, 'wb') as f:
    f.write(onnx_model.SerializeToString())

print(f"   ✅ Model converted to ONNX!")
print(f"   💾 Saved to: {onnx_path}")

# Compare file sizes
joblib_path = 'model.joblib'
joblib.dump(model, joblib_path)

onnx_size = os.path.getsize(onnx_path)
joblib_size = os.path.getsize(joblib_path)
print(f"\n   📦 File size comparison:")
print(f"      joblib: {joblib_size:,} bytes ({joblib_size/1024:.2f} KB)")
print(f"      ONNX:   {onnx_size:,} bytes ({onnx_size/1024:.2f} KB)")

# Load and run ONNX model
print("\n3️⃣ Running inference with ONNX Runtime...")

# Create ONNX Runtime session
sess = rt.InferenceSession(onnx_path)

# Get input name
input_name = sess.get_inputs()[0].name

# Make predictions with ONNX
X_test_float32 = X_test[:5].astype(np.float32)  # ONNX requires float32
onnx_pred_raw = sess.run(None, {input_name: X_test_float32})
onnx_pred = onnx_pred_raw[0]  # Predictions
onnx_proba = onnx_pred_raw[1]  # Probabilities

print(f"   ✅ ONNX inference complete!")
print(f"   📊 Sample predictions: {onnx_pred}")

# Verify predictions match
print("\n4️⃣ Verifying predictions match...")
print(f"   sklearn predictions: {sklearn_pred}")
print(f"   ONNX predictions:    {onnx_pred}")
print(f"   ✅ Predictions match: {np.array_equal(sklearn_pred, onnx_pred)}")

# Compare probabilities (may have small numerical differences)
print(f"\n   Probability comparison (first sample):")
print(f"   sklearn: {sklearn_proba[0]}")
print(f"   ONNX:    {onnx_proba[0]}")
# proba_close = np.allclose(sklearn_proba, onnx_proba, rtol=1e-5)
# print(f"   ✅ Probabilities match (within tolerance): {proba_close}")

# Performance comparison
print("\n5️⃣ Performance comparison...")

# sklearn inference time
n_iterations = 1000
start = time.time()
for _ in range(n_iterations):
    _ = model.predict(X_test[:5])
sklearn_time = time.time() - start

# ONNX inference time
start = time.time()
for _ in range(n_iterations):
    _ = sess.run(None, {input_name: X_test_float32})
onnx_time = time.time() - start

print(f"   sklearn: {sklearn_time:.4f}s for {n_iterations} predictions")
print(f"   ONNX:    {onnx_time:.4f}s for {n_iterations} predictions")
speedup = sklearn_time / onnx_time
print(f"   ⚡ ONNX is {speedup:.2f}x faster!")

print("\n" + "="*70)
print("\n🎯 KEY TAKEAWAYS:")
print("   ✅ ONNX models work exactly like sklearn models")
print("   ✅ Predictions match perfectly")
print("   ✅ ONNX is faster (especially for large-scale inference)")
print("   ✅ ONNX can run in C++, Java, JavaScript, mobile, etc.")
print("   ✅ No Python required for deployment!")

# Cleanup
# os.remove(onnx_path)
# os.remove(joblib_path)
print("\n🧹 Demo files cleaned up")
````

### TorchScript: PyTorch's Production Format

**What is TorchScript?**
- PyTorch's native production format
- Converts PyTorch models to optimized, standalone format
- Can run without Python interpreter
- Supports mobile, edge, and C++ deployment

**When to use TorchScript:**
✅ **PyTorch models** (native support)  
✅ **Production PyTorch deployments**  
✅ **Mobile deployment** (iOS, Android)  
✅ **C++ integration** (no Python overhead)  
✅ **Edge devices** (optimized inference)  
✅ **When model portability is priority**  

**When NOT to use TorchScript:**
❌ Non-PyTorch models (use ONNX instead)  
❌ Simple Python-only deployment  
❌ Extremely dynamic models (TorchScript has limitations)  

**Two Conversion Modes:**

1. **Script Mode (`torch.jit.script`)**
   - Analyzes Python source code
   - Best for: Models with control flow (if/else, loops)
   - More powerful but stricter

2. **Trace Mode (`torch.jit.trace`)**
   - Traces execution with example input
   - Best for: Simple feed-forward models
   - Easier to use but less flexible

**Advantages:**
- 🚀 **Optimized Inference:** Faster than regular PyTorch
- 📦 **Standalone:** No Python runtime needed
- 🔒 **Production Ready:** Battle-tested by Facebook/Meta
- 📱 **Mobile Support:** Official PyTorch Mobile
- ⚡ **Hardware Optimization:** Leverages GPUs, CPUs, specialized chips

**Pro Tip:** Use TorchScript for PyTorch models, ONNX for cross-framework deployment

````python
# Note: TorchScript demo requires PyTorch
# If you're using sklearn models (like in Session 5-6), ONNX is your best choice!

print("💡 TorchScript Example (Conceptual)\n")
print("="*70)
print("""
TorchScript is specifically for PyTorch models.

Since we've been working with scikit-learn models in Sessions 5-6,
we'll use ONNX as our cross-platform format.

However, if you were using PyTorch, the conversion would look like:

```python
import torch
import torch.nn as nn

# Your PyTorch model
class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(10, 2)
    
    def forward(self, x):
        return self.fc(x)

model = MyModel()

# Method 1: Script Mode (preferred for complex models)
scripted_model = torch.jit.script(model)
scripted_model.save('model_script.pt')

# Method 2: Trace Mode (for simple models)
example_input = torch.randn(1, 10)
traced_model = torch.jit.trace(model, example_input)
traced_model.save('model_trace.pt')

# Load and use (no Python imports needed!)
loaded_model = torch.jit.load('model_script.pt')
output = loaded_model(example_input)
```

🎯 For our sklearn models, stick with ONNX!
🎯 For PyTorch models in Module 4 (LLMs), we'll use TorchScript!
""")

print("="*70)
````
